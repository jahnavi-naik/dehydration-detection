```python
from pathlib import Path
import pandas as pd

data_dir = Path("shimmer-raw-dataset-dehydration")
```

```python
files = [f for f in data_dir.rglob("*") if f.is_file()]

print("Total files:", len(files))

for f in files[:50]:
    print(f.relative_to(data_dir))
```

```python
csv_files = list(data_dir.rglob("*.csv"))

print("Number of CSV files:", len(csv_files))
print("First few files:", csv_files[:10])

if csv_files:
    df = pd.read_csv(csv_files[0], nrows=1000)

    print("Shape:", df.shape)
    print("Columns:", df.columns.tolist())

    display(df.head())
```

```python
import pandas as pd

# Read the first file
file_path = csv_files[0]

df = pd.read_csv(
    file_path,
    header=1,
    skiprows=[2, 3],
    low_memory=False
)

print(df.shape)
print(df.columns.tolist())
display(df.head())
```

```python
# Read the first four rows without assigning headers
metadata = pd.read_csv(
    file_path,
    header=None,
    nrows=4
)

# Row 1: sensor names
sensor_names = metadata.iloc[1]

# Row 2: calibration status
calibration = metadata.iloc[2]

# Row 3: units
units = metadata.iloc[3]

# Inspect available sensors
sensor_info = pd.DataFrame({
    "Sensor": sensor_names,
    "Calibration": calibration,
    "Units": units
})

display(sensor_info)
```

```python
# Select calibrated columns
calibrated_indices = sensor_info.index[
    sensor_info["Calibration"] == "CAL"
].tolist()

# Keep the original column positions
df = pd.read_csv(
    file_path,
    header=None,
    skiprows=4,
    usecols=calibrated_indices
)

# Assign sensor names
df.columns = [
    f"{sensor_names[i]}_{units[i]}"
    for i in calibrated_indices
]

# Convert to numeric values
df = df.apply(pd.to_numeric, errors="coerce")

print("Shape:", df.shape)
print("Columns:", df.columns.tolist())

display(df.head())
```

```python
print("Dataset shape:", df.shape)

missing = df.isna().sum().sort_values(ascending=False)

display(missing)

display(df.describe())
```

```python
import matplotlib.pyplot as plt

# Find relevant sensor columns
sensor_keywords = ["PPG", "GSR", "Accel"]

for keyword in sensor_keywords:
    matching = [
        col for col in df.columns
        if keyword.lower() in col.lower()
    ]

    if matching:
        col = matching[0]

        plt.figure(figsize=(12, 3))
        plt.plot(df[col].iloc[:2000])
        plt.title(f"{keyword}: {col}")
        plt.xlabel("Sample index")
        plt.ylabel(col)
        plt.show()
```

```python
import pandas as pd

# Original column positions
selected_columns = {
    0: "PPG",
    3: "Accel_X",
    4: "Accel_Y",
    7: "Accel_Z",
    18: "Temperature",
    20: "GSR_Resistance",
    21: "Gyro_Z",
    25: "Timestamp",
    27: "Gyro_X",
    30: "Gyro_Y",
    31: "GSR_Conductance"
}

df = pd.read_csv(
    csv_files[0],
    header=None,
    skiprows=4,
    usecols=list(selected_columns.keys())
)

# Rename columns using their original positions
df.columns = [
    selected_columns[i] for i in sorted(selected_columns)
]

# Convert all columns to numeric
df = df.apply(pd.to_numeric, errors="coerce")

print("Shape:", df.shape)
display(df.head())
```

```python
print("Number of rows:", len(df))

print("\nMissing values:")
print(df.isna().sum())

print("\nSummary statistics:")
display(df.describe())
```

```python
df["Time_seconds"] = (
    df["Timestamp"] - df["Timestamp"].iloc[0]
) / 1000

print("Recording duration (minutes):")
print(df["Time_seconds"].iloc[-1] / 60)
```

```python
import matplotlib.pyplot as plt

signals = [
    "PPG",
    "GSR_Conductance",
    "Temperature"
]

fig, axes = plt.subplots(3, 1, figsize=(12, 9))

# Plot the first 10,000 samples
sample = df.iloc[:10000]

for ax, signal in zip(axes, signals):
    ax.plot(sample["Time_seconds"], sample[signal])
    ax.set_title(f"{signal} Over Time")
    ax.set_xlabel("Time (seconds)")
    ax.set_ylabel(signal)
    ax.grid(alpha=0.3)

plt.tight_layout()
plt.show()
```

```python
df["Accel_Magnitude"] = np.sqrt(
    df["Accel_X"]**2 +
    df["Accel_Y"]**2 +
    df["Accel_Z"]**2
)
```

```python
# Assign each sample to a 60-second window
df["window"] = (df["Time_seconds"] // 60).astype(int)

features = df.groupby("window").agg(
    PPG_mean=("PPG", "mean"),
    PPG_std=("PPG", "std"),
    GSR_mean=("GSR_Conductance", "mean"),
    GSR_std=("GSR_Conductance", "std"),
    Temp_mean=("Temperature", "mean"),
    Temp_std=("Temperature", "std"),
    Accel_mean=("Accel_Magnitude", "mean"),
    Accel_std=("Accel_Magnitude", "std")
).reset_index()

display(features.head())
```
